**STEP 1 : Data Understanding & Cleaning**

```python
# 1.1 Handling Missing Values

# Columns with ~50% missing values
partial_missing_cols = [
    'Malware Indicators', 'Alerts/Warnings',
    'Proxy Information', 'Firewall Logs', 'IDS/IPS Alerts'
]

# Check unique non-null values for each column
for col in partial_missing_cols:
    unique_values = df[col].dropna().unique()
    print(f"{col}: {len(unique_values)} unique non-null values")
    print(f"Sample values: {unique_values[:3]}\n")

# Drop the ones with only 1 unique non-null value
cols_to_drop = ['Malware Indicators', 'Alerts/Warnings', 'Firewall Logs', 'IDS/IPS Alerts','Payload Data']


# Create a new dataframe with selected columns dropped
df_cleaned = df.drop(columns=cols_to_drop)

print(f"\nDropped columns: {cols_to_drop}")
print(f"New DataFrame shape: {df_cleaned.shape}")
```

```python
# 1.2 Imputations

# Impute categorical columns with mode
for col in df_cleaned.select_dtypes(include='object').columns:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col].fillna(df_cleaned[col].mode()[0], inplace=True)

# Impute numeric columns with median
for col in df_cleaned.select_dtypes(include='number').columns:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col].fillna(df_cleaned[col].median(), inplace=True)

# Extract time-based features
df_cleaned['hour'] = df_cleaned['Timestamp'].dt.hour
df_cleaned['day'] = df_cleaned['Timestamp'].dt.day
df_cleaned['is_nighttime'] = df_cleaned['hour'].apply(lambda x: 1 if (0 <= x < 6) else 0)

# Final check
print("\n➡️ Remaining missing values (after cleaning):")
display(df_cleaned.isnull().sum())

print("\n➡️ Final shape of cleaned dataset:", df_cleaned.shape)
```

**STEP 2 : Feature Engineering**

```python
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer

df_fe = df_cleaned.copy()

# ---------------------------------------------
# 🔹 2.1 Categorical Encoding: Frequency Encoding
# ---------------------------------------------

categorical_cols = ['Protocol', 'Packet Type', 'Traffic Type', 'Action Taken', 'Severity Level']
for col in categorical_cols:
    freq_encoding = df_fe[col].value_counts().to_dict()
    df_fe[f'{col}_freq'] = df_fe[col].map(freq_encoding)

# --------------------------------------------------
# 🔹 2.2 Encode Source/Destination IP using Hashing
# --------------------------------------------------

df_fe['Source_IP_hash'] = df_fe['Source IP Address'].apply(lambda x: hash(x) % 10**6)
df_fe['Destination_IP_hash'] = df_fe['Destination IP Address'].apply(lambda x: hash(x) % 10**6)

# ---------------------------------------------
# 🔹 2.3 Text Features
# ---------------------------------------------

#Encoding Attack Signauture Column since it has only 2 unique values
df_fe['Attack Signature'] = df_fe['Attack Signature'].map({
    'Known Pattern A': 0,
    'Known Pattern B': 1
})

# -----------------------------------------------
# 🔹 2.5 Normalize/Standardize Numeric Columns
# -----------------------------------------------

numeric_cols = ['Packet Length', 'Anomaly Scores', 'hour', 'day']
scaler = StandardScaler()
df_fe[numeric_cols] = scaler.fit_transform(df_fe[numeric_cols])

# Final output
print("✅ Feature engineering complete.")
print("🧾 Final shape after feature engineering:", df_fe.shape)
```

**STEP 3 : Dimensionality Reduction**

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Step 1: Sample the dataset (for visualization only)
df_sample = df_fe.sample(n=3000, random_state=42)

# Step 2: Keep only numeric features
numeric_sample = df_sample.select_dtypes(include='number')

# Step 3: Apply PCA to reduce dimensionality
n_components = min(numeric_sample.shape[1], 30)
pca = PCA(n_components=n_components, random_state=42)
pca_result = pca.fit_transform(numeric_sample)

# Step 4: Apply t-SNE to reduce to 2D
tsne = TSNE(n_components=2, random_state=42, perplexity=30, max_iter=500)
tsne_result = tsne.fit_transform(pca_result)

# Step 5: Add t-SNE results to sample DataFrame
df_sample['tsne-2d-one'] = tsne_result[:, 0]
df_sample['tsne-2d-two'] = tsne_result[:, 1]

# Step 6: Plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='tsne-2d-one', y='tsne-2d-two',
    hue='Attack Type',
    palette='tab10',
    data=df_sample,
    legend='full',
    alpha=0.5,
    s=12
)
plt.title("🔍 t-SNE Visualization (on Sampled PCA-reduced Data)")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load the dataset
df = pd.read_csv('cybersecurity_attacks.csv')

# --- Data Preprocessing ---
# Select relevant numerical features
numerical_features = df[['Anomaly Scores', 'Packet Length', 'Source Port', 'Destination Port']]

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(numerical_features)

# --- Dimensionality Reduction for 3D Visualization ---
# Reduce to 3 components for 3D plotting
pca_3d = PCA(n_components=3)
X_pca_3d = pca_3d.fit_transform(X_scaled)

# --- Visualization before KMeans (3D) ---
fig_before = plt.figure(figsize=(10, 8))
ax_before = fig_before.add_subplot(111, projection='3d')
ax_before.scatter(X_pca_3d[:, 0], X_pca_3d[:, 1], X_pca_3d[:, 2], alpha=0.5)
ax_before.set_title('Data Distribution Before K-Means (3D PCA)')
ax_before.set_xlabel('Principal Component 1')
ax_before.set_ylabel('Principal Component 2')
ax_before.set_zlabel('Principal Component 3')
plt.savefig('before_kmeans_3d.png')
plt.show()


# --- Applying KMeans (using the same parameters as before) ---
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)

# --- Visualization after KMeans (3D) ---
fig_after = plt.figure(figsize=(12, 10))
ax_after = fig_after.add_subplot(111, projection='3d')
scatter_after = ax_after.scatter(X_pca_3d[:, 0], X_pca_3d[:, 1], X_pca_3d[:, 2], c=cluster_labels, cmap='viridis', alpha=0.5)
ax_after.set_title('Data Distribution After K-Means (3D PCA, k=3)')
ax_after.set_xlabel('Principal Component 1')
ax_after.set_ylabel('Principal Component 2')
ax_after.set_zlabel('Principal Component 3')
legend1 = ax_after.legend(*scatter_after.legend_elements(), title="Clusters")
ax_after.add_artist(legend1)
plt.savefig('after_kmeans_3d.png')
plt.show()
# --- Visualization 1: Feature-based 3D plot (Before KMeans) ---
fig_before = plt.figure(figsize=(12, 10))
ax_before = fig_before.add_subplot(111, projection='3d')

# Scatter plot using original features, all in one color
ax_before.scatter(df['Anomaly Scores'], df['Packet Length'], df['Source Port'], alpha=0.5)

ax_before.set_title('Data Distribution Before Clustering (by Features)')
ax_before.set_xlabel('Anomaly Scores')
ax_before.set_ylabel('Packet Length')
ax_before.set_zlabel('Source Port')
plt.savefig('before_kmeans_3d_features.png')
plt.show()


# --- Visualization 2: Feature-based 3D plot (After KMeans) ---
fig_after = plt.figure(figsize=(12, 10))
ax_after = fig_after.add_subplot(111, projection='3d')

# Scatter plot using original features, colored by cluster
scatter_after = ax_after.scatter(df['Anomaly Scores'], df['Packet Length'], df['Source Port'], c=cluster_labels, cmap='viridis', alpha=0.5)

ax_after.set_title('Clusters by Anomaly Score, Packet Length, and Source Port')
ax_after.set_xlabel('Anomaly Scores')
ax_after.set_ylabel('Packet Length')
ax_after.set_zlabel('Source Port')
legend = ax_after.legend(*scatter_after.legend_elements(), title="Clusters")
ax_after.add_artist(legend)
plt.savefig('after_kmeans_3d_features.png')
plt.show()
```

```python

from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler
# --- Data Preprocessing ---
# Select relevant numerical features for the model
numerical_features = df[['Anomaly Scores', 'Packet Length', 'Source Port', 'Destination Port']]

# SVMs are sensitive to feature scaling, so we standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(numerical_features)

# --- Applying One-Class SVM ---
# The 'nu' parameter is an upper bound on the fraction of training errors
# and a lower bound of the fraction of support vectors. It's essentially
# the expected proportion of anomalies in the data.
one_class_svm = OneClassSVM(nu=0.05) # Assuming ~5% of the data are anomalies
predictions = one_class_svm.fit_predict(X_scaled)

# Add predictions to the dataframe (-1 for anomalies, 1 for inliers)
df['Anomaly'] = predictions

# --- 2D VISUALIZATIONS ---

# 1. Before One-Class SVM (2D)
plt.figure(figsize=(10, 7))
plt.scatter(df['Anomaly Scores'], df['Packet Length'], alpha=0.6,)
plt.title('Data Distribution Before One-Class SVM (2D)')
plt.xlabel('Anomaly Scores')
plt.ylabel('Packet Length')
plt.grid(True)
plt.savefig('before_one_class_svm_2d.png')
plt.show()

# 2. After One-Class SVM (2D)
plt.figure(figsize=(10, 7))
inliers = df[df['Anomaly'] == 1]
outliers = df[df['Anomaly'] == -1]
plt.scatter(inliers['Anomaly Scores'], inliers['Packet Length'], c='deepskyblue', label='Normal', alpha=0.6)
plt.scatter(outliers['Anomaly Scores'], outliers['Packet Length'], c='#ff7f0e', label='Anomaly', alpha=0.6) # Bright Orange
plt.title('Data Distribution After One-Class SVM (2D)')
plt.xlabel('Anomaly Scores')
plt.ylabel('Packet Length')
plt.legend()
plt.grid(True)
plt.savefig('after_one_class_svm_2d.png')
plt.show()


# --- 3D VISUALIZATIONS ---

# 3. Before One-Class SVM (3D)
fig_before_3d = plt.figure(figsize=(12, 10))
ax_before_3d = fig_before_3d.add_subplot(111, projection='3d')
ax_before_3d.scatter(df['Anomaly Scores'], df['Packet Length'], df['Source Port'], alpha=0.5, )
ax_before_3d.set_title('Data Distribution Before One-Class SVM (3D)')
ax_before_3d.set_xlabel('Anomaly Scores')
ax_before_3d.set_ylabel('Packet Length')
ax_before_3d.set_zlabel('Source Port')
plt.savefig('before_one_class_svm_3d.png')
plt.show()

# 4. After One-Class SVM (3D)
fig_after_3d = plt.figure(figsize=(12, 10))
ax_after_3d = fig_after_3d.add_subplot(111, projection='3d')
inliers_3d = df[df['Anomaly'] == 1]
outliers_3d = df[df['Anomaly'] == -1]
ax_after_3d.scatter(inliers_3d['Anomaly Scores'], inliers_3d['Packet Length'], inliers_3d['Source Port'], c='deepskyblue', label='Normal', alpha=0.5)
ax_after_3d.scatter(outliers_3d['Anomaly Scores'], outliers_3d['Packet Length'], outliers_3d['Source Port'], c='#ff7f0e', label='Anomaly', alpha=0.5) # Bright Orange
ax_after_3d.set_title('Data Distribution After One-Class SVM (3D)')
ax_after_3d.set_xlabel('Anomaly Scores')
ax_after_3d.set_ylabel('Packet Length')
ax_after_3d.set_zlabel('Source Port')
ax_after_3d.legend()
plt.savefig('after_one_class_svm_3d.png')
plt.show()
```
